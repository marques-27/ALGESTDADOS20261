```python
#16 tempo de atendimento

soma = 0
contador = 0
tempo = float(input('digite_o_tempo_gasto(0_para_encerrar):'))
while tempo != 0:
    soma += tempo
    contador += 1
    tempo = float(input('digite_o_tempo_gasto(0_para_encerrar):'))

if contador > 0:
    media = soma / contador
    print(f'total de clientes:{contador}')
    print(f'tempo medio:{media:.2f}')
else:
    print('ninguém foi atendido')
```

```python
numnegativos = 0
numpositivos = 0


numero = int(input(f'digite o 1º número: '))
maior = numero


if numero < 0:
    numnegativos += 1
elif numero > 0:
    numpositivos += numero


for i in range(2, 11):
    numero = int(input(f'digite o {i}º número: '))

    if numero > maior:
        maior = numero

    if numero < 0:
        numnegativos += 1
    elif numero > 0:
        numpositivos += numero


print(f'\n Resultados Finais ')
print(f'Total de números negativos: {numnegativos}')
print(f'Soma dos positivos: {numpositivos}')
print(f'Maior valor: {maior}')
```
