```python
# 16) Tempo de Atendimento
# Calcula o tempo médio de atendimento dos clientes

soma = 0
contador = 0

tempo = float(input('Digite o tempo de atendimento (0 para encerrar): '))

while tempo != 0:
    soma += tempo
    contador += 1
    tempo = float(input('Digite o tempo de atendimento (0 para encerrar): '))

if contador > 0:
    media = soma / contador
    print(f'\n=== RESULTADO ===')
    print(f'Total de clientes: {contador}')
    print(f'Tempo médio: {media:.2f} minutos')
else:
    print('Ninguém foi atendido')
```

```python
# 17) Análise de 10 Números
# Identifica negativos, soma positivos e encontra o maior

numnegativos = 0
numpositivos = 0

numero = int(input('Digite o 1º número: '))
maior = numero

if numero < 0:
    numnegativos += 1
elif numero > 0:
    numpositivos += numero

for i in range(2, 11):
    numero = int(input(f'Digite o {i}º número: '))

    if numero > maior:
        maior = numero

    if numero < 0:
        numnegativos += 1
    elif numero > 0:
        numpositivos += numero

print(f'\n=== RESULTADO ===')
print(f'Total de números negativos: {numnegativos}')
print(f'Soma dos positivos: {numpositivos}')
print(f'Maior valor: {maior}')
```

```python
# 18) Consumo de Energia
# Calcula consumo mensal e valor a pagar com tarifa

soma = 0
tarifa = 0.75  # R$ por kWh

for mes in range(1, 13):
    consumo = float(input(f'Consumo do mês {mes} (kWh): '))
    soma += consumo

media = soma / 12
valor = soma * tarifa

print(f'\n=== RESULTADO ===')
print(f'Consumo total anual: {soma:.2f} kWh')
print(f'Consumo médio mensal: {media:.2f} kWh')
print(f'Valor a pagar (R$ {tarifa}/kWh): R$ {valor:.2f}')
```

```python
# 19) Adivinhe o Número Secreto
# Jogo simples de adivinhar um número com limite de tentativas

import random

numero_secreto = random.randint(1, 10)
tentativas = 0
acertou = False

while tentativas < 3 and not acertou:
    numero = int(input('Adivinhe o número (1 a 10): '))
    tentativas += 1
    
    if numero == numero_secreto:
        print(f'✓ Parabéns! Acertou em {tentativas} tentativa(s)!')
        acertou = True
    elif numero < numero_secreto:
        print('O número é maior')
    else:
        print('O número é menor')

if not acertou:
    print(f'✗ Fim do jogo! O número secreto era {numero_secreto}')
```

```python
# 20) Estatísticas de Notas
# Calcula média, maior e menor nota da turma

soma = 0
contador = 0
maior_nota = 0
menor_nota = 10

while True:
    nota = float(input('Digite a nota (ou -1 para encerrar): '))
    
    if nota == -1:
        break
    
    if nota < 0 or nota > 10:
        print('Nota inválida! Digite entre 0 e 10.')
        continue
    
    soma += nota
    contador += 1
    
    if nota > maior_nota:
        maior_nota = nota
    if nota < menor_nota:
        menor_nota = nota

if contador > 0:
    media = soma / contador
    print(f'\n=== RESULTADO ===')
    print(f'Total de notas: {contador}')
    print(f'Média: {media:.2f}')
    print(f'Maior nota: {maior_nota:.2f}')
    print(f'Menor nota: {menor_nota:.2f}')
else:
    print('Nenhuma nota foi registrada')
```

```python
# 21) Pesquisa Eleitoral
# Contagem de votos para 3 candidatos

candidato_a = 0
candidato_b = 0
candidato_c = 0

while True:
    voto = int(input('Vote (1-Candidato A, 2-Candidato B, 3-Candidato C, 0-Encerrar): '))
    
    if voto == 0:
        break
    elif voto == 1:
        candidato_a += 1
    elif voto == 2:
        candidato_b += 1
    elif voto == 3:
        candidato_c += 1
    else:
        print('Voto inválido!')

total = candidato_a + candidato_b + candidato_c

print(f'\n=== RESULTADO ELEITORAL ===')
print(f'Candidato A: {candidato_a} votos')
print(f'Candidato B: {candidato_b} votos')
print(f'Candidato C: {candidato_c} votos')
print(f'Total de votos: {total}')

if total > 0:
    if candidato_a > candidato_b and candidato_a > candidato_c:
        print('Vencedor: Candidato A')
    elif candidato_b > candidato_a and candidato_b > candidato_c:
        print('Vencedor: Candidato B')
    elif candidato_c > candidato_a and candidato_c > candidato_b:
        print('Vencedor: Candidato C')
    else:
        print('Empate entre candidatos')
```

```python
# 22) Estacionamento
# Calcula valor de estacionamento por hora

horas = int(input('Quantas horas o carro ficou estacionado? '))
valor_hora = 5.00  # R$ por hora
valor_total = horas * valor_hora

# Desconto para estacionamentos longos (acima de 8 horas)
if horas > 8:
    desconto = valor_total * 0.10  # 10% de desconto
    valor_total -= desconto
    print(f'\n=== RESULTADO ===')
    print(f'Valor original: R$ {horas * valor_hora:.2f}')
    print(f'Desconto (10%): R$ {desconto:.2f}')
    print(f'Valor a pagar: R$ {valor_total:.2f}')
else:
    print(f'\n=== RESULTADO ===')
    print(f'Horas: {horas}')
    print(f'Valor a pagar: R$ {valor_total:.2f}')
```

```python
# 23) Divisíveis por 2 e por 3
# Conta números divisíveis por 2, por 3 e por ambos

div_2 = 0
div_3 = 0
div_ambos = 0

for i in range(1, 101):
    if i % 2 == 0 and i % 3 == 0:
        div_ambos += 1
    elif i % 2 == 0:
        div_2 += 1
    elif i % 3 == 0:
        div_3 += 1

print(f'=== NÚMEROS DE 1 A 100 ===')
print(f'Divisíveis apenas por 2: {div_2}')
print(f'Divisíveis apenas por 3: {div_3}')
print(f'Divisíveis por 2 E por 3: {div_ambos}')
print(f'Total (2 ou 3): {div_2 + div_3 + div_ambos}')
```

```python
# 24) Classificação Escolar
# Classifica alunos por conceito baseado em nota

soma = 0
total_alunos = 0
conceito_a = 0  # nota >= 9
conceito_b = 0  # nota >= 7 e < 9
conceito_c = 0  # nota >= 5 e < 7
conceito_d = 0  # nota < 5

while True:
    nota = float(input('Digite a nota do aluno (ou -1 para encerrar): '))
    
    if nota == -1:
        break
    
    if nota < 0 or nota > 10:
        print('Nota inválida!')
        continue
    
    soma += nota
    total_alunos += 1
    
    if nota >= 9:
        conceito_a += 1
    elif nota >= 7:
        conceito_b += 1
    elif nota >= 5:
        conceito_c += 1
    else:
        conceito_d += 1

if total_alunos > 0:
    media = soma / total_alunos
    print(f'\n=== CLASSIFICAÇÃO ===')
    print(f'Total de alunos: {total_alunos}')
    print(f'Média da turma: {media:.2f}')
    print(f'Conceito A (9.0-10.0): {conceito_a}')
    print(f'Conceito B (7.0-8.9): {conceito_b}')
    print(f'Conceito C (5.0-6.9): {conceito_c}')
    print(f'Conceito D (0.0-4.9): {conceito_d}')
else:
    print('Nenhum aluno foi registrado')
```

```python
# 25) Análise de Vendas
# Calcula total de vendas, maior venda e desconto aplicado

total_vendas = 0
maior_venda = 0
contador = 0
com_desconto = 0

while True:
    valor = float(input('Digite o valor da venda (ou 0 para encerrar): '))
    
    if valor == 0:
        break
    
    contador += 1
    
    # Aplica 5% de desconto para vendas acima de R$ 500
    if valor > 500:
        desconto = valor * 0.05
        valor_final = valor - desconto
        com_desconto += 1
    else:
        valor_final = valor
    
    total_vendas += valor_final
    
    if valor_final > maior_venda:
        maior_venda = valor_final

if contador > 0:
    media = total_vendas / contador
    print(f'\n=== ANÁLISE DE VENDAS ===')
    print(f'Total de vendas: {contador}')
    print(f'Valor total arrecadado: R$ {total_vendas:.2f}')
    print(f'Média por venda: R$ {media:.2f}')
    print(f'Maior venda: R$ {maior_venda:.2f}')
    print(f'Vendas com desconto: {com_desconto}')
else:
    print('Nenhuma venda foi registrada')
```
