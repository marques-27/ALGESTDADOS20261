```python
#16 tempo de atendimento

soma = 0
contador = 0
tempo = float(input('digite_o_tempo_gasto(0_para_encerrar):'))
while tempo != 0:
    soma += tempo
    contador += 1
    tempo = float(input('digite_o_tempo_gasto(0_para_encerrar):'))

if contador > 0:
    media = soma / contador
    print(f'total de clientes:{contador}')
    print(f'tempo medio:{media:.2f}')
else:
    print('ninguém foi atendido')
```

```python
#17 análise de 10 números

numnegativos = 0
numpositivos = 0


numero = int(input(f'digite o 1º número: '))
maior = numero


if numero < 0:
    numnegativos += 1
elif numero > 0:
    numpositivos += numero


for i in range(2, 11):
    numero = int(input(f'digite o {i}º número: '))

    if numero > maior:
        maior = numero

    if numero < 0:
        numnegativos += 1
    elif numero > 0:
        numpositivos += numero


print(f'\n Resultados Finais ')
print(f'Total de números negativos: {numnegativos}')
print(f'Soma dos positivos: {numpositivos}')
print(f'Maior valor: {maior}')
```

```python
# 21 pesquisa eleitoral

candidato_A = 0
candidato_B = 0
candidato_C = 0

while True:
    voto = int(input('Vote(1-candidato_A, 2-candidato_B,3-candidato_C,0-encerrar votos):'))
    if voto == 0:
        break
    elif voto == 1:
        candidato_A += 1
    elif voto == 2:
        candidato_B += 1
    elif voto == 3:
        candidato_C += 1
    else:
        print('voto inválido!')

total = candidato_A  + candidato_B  + candidato_C
print(f'\n resultado eleitoral')
print(f'candidato_A: {candidato_A} votos')
print(f'candidato_B: {candidato_B} votos')
print(f'candidato_C: {candidato_C} votos')
print(f'total de votos: {total}')

if total > 0:
    if candidato_A > candidato_B and candidato_A > candidato_C:
        print('vencedor: candidato_A')
    elif candidato_B > candidato_A and candidato_B > candidato_C:
        print('vencedor: candidato_B')
    elif candidato_C > candidato_A  and candidato_C > candidato_B:
        print('vencedor: candidato_C')
    else:
        print('empate entre candidatos')
```

```python
# 22  estacionamento

total_arrecadado = 0

while True:
    horas = int(input('quantas horas o carro ficou estacionado?(0 para encerrar):'))

    if horas == 0:
        break

    valor_hora = 5.00
    valor_total = horas * valor_hora

    print(f'\n resultado')

    if horas > 5 :
        desconto = valor_total * 0.10
        valor_total -= desconto
        print(f'valor original: R$ {horas * valor_hora:.2f} ')
        print(f'desconto(10%): R$ {desconto:.2f} ')
        print(f'valor a pagar: R$ {valor_total:.2f}')
    else:
        print(f'horas: {horas}')
        print(f'valor a pagar: R$ {valor_total:.2f}')

    total_arrecadado += valor_total

print(f'\nTotal arrecadado no dia: R$ {total_arrecadado:.2f}')
```

```python
# 28 caça aos primos

total_primos = 0
soma_nao_primos = 0

print('Digite números inteiros. Digite -1 para encerrar.')

while True:
    numero_digitado = int(input('Digite um número: '))
    if numero_digitado == -1:
        break


    eh_primo = True
    if numero_digitado <= 1:
        eh_primo = False
    else:

        for i in range(2, numero_digitado):
            if numero_digitado % i == 0:
                eh_primo = False
                break

    if eh_primo:
        total_primos += 1
    else:
        soma_nao_primos += numero_digitado

print(f'\nResultados Finais:')
print(f'Total de números primos: {total_primos}')
print(f'Soma dos números não primos: {soma_nao_primos}')
```

```python

#30 calculo de fatorial

numero = int(input("digite um numero positivo para o calculo do fatorial"))
fatorial = 1

if numero < 0:
    print("Fatorial não existe para números negativos.")
elif numero == 0:
    print("O fatorial de 0 é 1")
    # Fatorial de 0 é 1 (ímpar)
    print("O fatorial é ímpar.")
elif numero > 0:
    for i in range(1, numero + 1):
        fatorial = fatorial * i
    print(f"O fatorial de {numero} é {fatorial}")

    if fatorial % 2 == 0:
        print("O fatorial é par.")
    else:
        print("O fatorial é ímpar.")
else:
    print("Por favor, digite um número inteiro válido.")
```
